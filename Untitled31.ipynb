```python
# fazer um programa que receba a idade de um aluno e classique ele em maior ou menor de idade

# entrada

idade = int(input("Digite sua idade: "))

# Processamento/Saída

if idade >= 18:
  print("Você é maior de idade")

else :
  print("Você é menor de idade")

while idade <0:
  print("Por favor, digite uma idade válida")
```

```python
# escrever um programa que recebe um número menor que 20 e imprime os números pares

# entrada

número = int(input("digite um número menor que 20: "))

# processamento / saída

if número < 20 and número >= 0 :
  for i in range(0,número + 1, 2):
     print(i)
else:
  print("Por favor, digite um número menor que 20 e maior ou igual a 0.")
```
