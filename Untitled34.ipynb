```python
# escrever um programa que recebe um número e imprima a contagem até o número

# entrada

número = int(input("Digite um número: "))

# processamento / saida

contador = 1
while contador <= número:
  print( contador )
  contador += 1
```

```python
# Faça um programa que solicite ao usuário 5 números solicitados e apresente a soma dos valores solicitados

soma = 0
for i in range(5):
    numero = int(input(f"Digite o {i+1}º número: "))
    soma += numero

print(f"A soma dos números é: {soma}")
```

```python
# Criar uma lista com 10 nomes de alunos. Em seguida, mostre

# * a lista completa
# * o terceiro nome
# * o ultimo nome

alunos = ["Alice", "Bruno", "Carla", "Daniel", "Eduarda", "Felipe", "Gabriela", "Hugo", "Isabela", "João"]

print("Lista completa de alunos:")
print(alunos)

print("\nTerceiro nome:")
print(alunos[2])

print("\nÚltimo nome:")
print(alunos[-1])
```

```python
# criar uma lista com cinco cores. Substitua a terceira cor por outra escolhida pelo usuário

cores = ["vermelho", "azul", "verde", "amarelo", "roxo"]
print("Lista original de cores:", cores)

nova_cor = input("Digite uma nova cor para substituir a terceira: ")

cores[2] = nova_cor

print("Lista de cores atualizada:", cores)
```

```python
# cadastro de nomes:
# crie uma lista vazia. Solicite tres nomes ao usuário e adicione-os na lista utilizando append()

nomes_cadastrados = []

for i in range(3):
    nome = input(f"Digite o {i+1}º nome para adicionar à lista: ")
    nomes_cadastrados.append(nome)

print("Nomes cadastrados:", nomes_cadastrados)
```
